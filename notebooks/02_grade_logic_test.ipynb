## **Step 1: Represent Grade Information**

We convert the grade information from the university regulations
into structured data so Python can process it reliably.

```python
grades = {
	"A": {"coefficient": 4.00, "status": "passing"},
	"A-": {"coefficient": 3.70, "status": "passing"},
	"B+": {"coefficient": 3.30, "status": "passing"},
	"B": {"coefficient": 3.00, "status": "passing"},
	"B-": {"coefficient": 2.70, "status": "passing"},
	"C+": {"coefficient": 2.30, "status": "passing"},
	"C": {"coefficient": 2.00, "status": "passing"},
	"C-": {"coefficient": 1.70, "status": "passing"},
	"D+": {"coefficient": 1.30, "status": "passing"},
	"D": {"coefficient": 1.00, "status": "passing"},
	"D-": {"coefficient": 0.70, "status": "failing"},
	"F": {"coefficient": 0.00, "status": "failing"},
}
```

## **Step 2:  Test it**

## **Step 3: Find the Lowest Passing Grade**

Create a reusable function that filters passing grades
and identifies the one with the lowest coefficient.

```python
def get_lowest_passing_grade(grades):
	passing_grades = {
		grade: information["coefficient"]
		for grade, information in grades.items()
		if information["status"] == "passing"
	}

	lowest_grade = min(
		passing_grades,
		key=passing_grades.get
	)

	return lowest_grade


result = get_lowest_passing_grade(grades)

print("Lowest passing grade:", result)
```

## **Step 4: Support Different Grade Questions**

Extend the rule-based logic to answer different types
of questions using the same structured grade data.

```python
def answer_grade_question(question, grades):
    question = question.lower()

    if "lowest passing" in question:
        return get_lowest_passing_grade(grades)

    elif "highest failing" in question:
        failing_grades = {
            grade: info["coefficient"]
            for grade, info in grades.items()
            if info["status"] == "failing"
        }

        return max(failing_grades, key=failing_grades.get)

    else:
        for grade in sorted(grades, key=len, reverse=True):
            if grade.lower() in question:

                if "failing" in question:
                    return (
                        "failing"
                        if grades[grade]["status"] == "failing"
                        else "passing"
                    )

                if any(word in question for word in [
                    "passing",
                    "pass",
                ]):
                    return grades[grade]["status"]

    return "I don't know."

    return "I don't know."
```

```python
questions = [
	"What is the lowest passing grade?",
	"What is the highest failing grade?",
	"Is D a passing grade?",
	"Is D- passing?",
]

for question in questions:
	answer = answer_grade_question(question, grades)
	print(f"Question: {question}")
	print(f"Answer: {answer}\n")
```

```python
test_cases = [
    ("What is the lowest passing grade?", "D"),
    ("What is the highest failing grade?", "D-"),
    ("Is A passing?", "passing"),
    ("Is C- passing?", "passing"),
    ("Is D+ passing?", "passing"),
    ("Is D- passing?", "failing"),
    ("Is F passing?", "failing"),
    ("Is B- passing?", "passing"),
    ("What is the lowest failing grade?", "I don't know."),
]

for question, expected in test_cases:
    actual = answer_grade_question(question, grades)

    status = "PASS" if actual == expected else "FAIL"

    print(f"{status} | Question: {question}")
    print(f"Expected: {expected} | Actual: {actual}\n")
```

```python
new_questions = [
    "Does D count as a passing grade?",
    "Can a student pass with D?",
    "Is D- a failing grade?",
    "Which grade has the lowest passing coefficient?",
]

for question in new_questions:
    answer = answer_grade_question(question, grades)
    print(f"Question: {question}")
    print(f"Answer: {answer}\n")
```

```python
edge_cases = [
    "Is D- passing or failing?",
    "Is D not a passing grade?",
    "Can a student pass with F?",
    "Does D+ count as a passing grade?",
    "Is D- lower than D?",
]

for question in edge_cases:
    answer = answer_grade_question(question, grades)
    print(f"Question: {question}")
    print(f"Answer: {answer}\n")
```
